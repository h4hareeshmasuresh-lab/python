```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------------
# STEP 1: Create Dataset (NumPy)
# -------------------------------
np.random.seed(10)

students = ["A", "B", "C", "D", "E", "F"]
Maths = np.random.randint(40, 100, 6)
Science = np.random.randint(40, 100, 6)

# Create DataFrame (Pandas)
df = pd.DataFrame({
    "Student": students,
    "Maths": Maths,
    "Science": Science
})

print("STUDENT DATASET:\n", df)

# -------------------------------
# STEP 2: LINE CHART
# -------------------------------
plt.figure()
plt.plot(df["Student"], df["Maths"], marker='o', label="Maths")
plt.plot(df["Student"], df["Science"], marker='s', label="Science")

plt.title("Student Marks Trend")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.grid(True)
plt.show()

# -------------------------------
# STEP 3: BAR CHART
# -------------------------------
plt.figure()
x = np.arange(len(students))

plt.bar(x - 0.2, df["Maths"], width=0.4, label="Maths")
plt.bar(x + 0.2, df["Science"], width=0.4, label="Science")

plt.xticks(x, students)
plt.title("Maths vs Science Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.show()

# -------------------------------
# STEP 4: SCATTER PLOT
# -------------------------------
plt.figure()

plt.scatter(
    df["Maths"],
    df["Science"],
    c=df["Science"],
    cmap='magma'
)

plt.colorbar(label="Science Marks")
plt.title("Maths vs Science Relationship")
plt.xlabel("Maths Marks")
plt.ylabel("Science Marks")
plt.show()

# -------------------------------
# STEP 5: PIE CHART
# -------------------------------
total_marks = df["Maths"]

plt.figure()
plt.pie(
    total_marks,
    labels=df["Student"],
    autopct='%1.1f%%'
)

plt.title("Maths Marks Distribution")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------------
# STEP 1: Create Dataset (NumPy)
# -------------------------------
np.random.seed(0)

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = np.random.randint(2000, 10000, 6)
profit = np.random.randint(500, 3000, 6)

# Create DataFrame (Pandas)
df = pd.DataFrame({
    "Month": months,
    "Sales": sales,
    "Profit": profit
})

print("DATASET:\n", df)

# -------------------------------
# STEP 2: LINE CHART (Sales Trend)
# -------------------------------
plt.figure()
plt.plot(df["Month"], df["Sales"], marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()

# -------------------------------
# STEP 3: BAR CHART (Sales vs Profit)
# -------------------------------
plt.figure()
x = np.arange(len(months))

plt.bar(x - 0.2, df["Sales"], width=0.4, label="Sales")
plt.bar(x + 0.2, df["Profit"], width=0.4, label="Profit")

plt.xticks(x, months)
plt.title("Sales vs Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()
plt.show()

# -------------------------------
# STEP 4: SCATTER PLOT (Relationship)
# -------------------------------
plt.figure()
plt.scatter(df["Sales"], df["Profit"], c=df["Profit"], cmap='magma')

plt.colorbar(label="Profit Level")
plt.title("Sales vs Profit Relationship")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()

# -------------------------------
# STEP 5: PIE CHART (Total Distribution)
# -------------------------------
total_sales = df["Sales"]

plt.figure()
plt.pie(total_sales, labels=df["Month"], autopct='%1.1f%%')
plt.title("Sales Distribution by Month")
plt.show()
```
