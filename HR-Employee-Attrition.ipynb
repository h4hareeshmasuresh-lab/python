### Part3-Data visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.CSV")

sns.countplot(x="Attrition",data=df)

plt.title("Employee Attrition Plot")
plt.xlabel("Attrition")
plt.ylabel("Number of employees")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.CSV")

sns.barplot(x="Department", y="EmployeeCount", data=df)

plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.title("Department_wise EmployeeCount")
plt.xticks(rotation=45)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.CSV")
plt.hist(df["Age"],
         bins=10,
         edgecolor="black")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.CSV")
sns.boxplot(y="MonthlyIncome",x="Attrition",data=df)
plt.title("Monthly income Vs Attrition")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.CSV")
sns.countplot(x="Attrition",data=df,hue="OverTime")
plt.title("Overtime Vs Attrition")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.CSV")
sns.heatmap(df.select_dtypes(["number"]).corr(),cmap="coolwarm")
plt.title("correlation Heatmap")

plt.show()
```
