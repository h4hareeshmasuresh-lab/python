# MATPLOTLIB – DATA VISUALIZATION IN PYTHON
### Introduction to Matplotlib
Matplotlib is one of the most popular Python libraries for data visualization. It allows users to create different types of graphs and charts to represent data in a simple and meaningful way.

Matplotlib supports various types of visualizations, such as:
+ Line plots
+ Bar charts
+ Scatter plots
+ Histograms
+ Pie charts
####  Installation of Matplotlib
To install Matplotlib, use the following command in the terminal or command prompt:

pip install matplotlib

##### Import Matplotlib into your Python program:

import matplotlib.pyplot as plt

Here, pyplot is a module used to create different types of graphs.
### Line Plot
A line plot is one of the most commonly used visualizations in data science. It is used to display data points connected by straight lines.

### Example 1: Basic Line Plot

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 12, 8, 4, 11]

plt.plot(x, y)

plt.title("Line Plot Example")
plt.xlabel("X-axis Label")
plt.ylabel("Y-axis Label")

plt.grid(True)#optinal grid
plt.show()
```

### Example 2:
### Line Plot – Weather Example
We can use a line plot to represent changes in temperature over different days.



```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [50, 51, 52, 48, 47, 46]

plt.plot(x, y, color="indigo",
         linewidth=3, linestyle=":")

plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Weather")

plt.grid(True)
plt.show()
```

### Explanation:
+ color – Changes the colour of the line.
+ linewidth – Controls the thickness of the line.
+ linestyle – Controls the style of the line.
+ grid(True) – Displays grid lines.
### Different Line Styles and Markers
Matplotlib provides different line styles and markers to make graphs more attractive and informative.

### Example 3:

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6, 7]
y = [80, 91, 92, 48, 47, 49, 40]

plt.plot(x, y, "g+:")

plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Weather")

plt.grid(True)
plt.show()
```

### Explanation:
+ g – Green colour.
+ + – Plus-shaped marker.
+ : – Dotted line style.
####  Customizing Line Colour, Marker and Line Style
We can customize the appearance of a line plot using different properties.

#### Example 4:

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [50, 51, 52, 48, 47, 49]

plt.plot(x, y, color="#00FF00",
         marker="*", linestyle="-")

plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Weather")

plt.grid(True)
plt.show()
```

### Marker Size Example

```python
plt.plot(x, y, 'b+-', markersize=10)
plt.show()
```

Here:
+ b – Blue colour.
+ + – Plus marker.
+ - – Solid line.
markersize=10 – Sets the marker size to 10.
###  Alpha Property
The alpha property controls the transparency of a line.

Its value ranges from 0 to 1.

+ alpha=0 – Completely transparent (invisible).
+ alpha=0.5 – 50% transparent.
+ alpha=1 – Completely visible (opaque).
#### Example:

```python
plt.plot(x, y, 'g<-', alpha=0.5)
plt.show()
```

### Labels, Grid and Legend
Labels, grid lines and legends help make graphs easier to understand.

A legend identifies different lines displayed in the same graph.

### Example 5: Weather Data Visualization

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]

max_t = [50, 51, 52, 43, 47, 49, 46]
min_t = [43, 42, 40, 44, 33, 35, 37]
avg_t = [45, 48, 48, 46, 40, 42, 41]

plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Weather")

plt.plot(days, max_t, label="Maximum")
plt.plot(days, min_t, label="Minimum")
plt.plot(days, avg_t, label="Average")

plt.legend(loc="best", shadow=True,
           fontsize="large")

plt.grid(True)
plt.show()
```

### Legend Properties
+ label – Assigns a name to each line.
+ plt.legend() – Displays the legend.
+ loc – Specifies the position of the legend.
+ shadow=True – Adds a shadow effect.
+ fontsize – Changes the legend font size.
#### Different Legend Locations:

+ upper right
+ upper left
+ lower right
+ lower left
+ best
#### Example:

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5, 6, 7]

max_t = [50, 51, 52, 43, 47, 49, 46]
min_t = [43, 42, 40, 44, 33, 35, 37]
avg_t = [45, 48, 48, 46, 40, 42, 41]

plt.xlabel("Day")
plt.ylabel("Temperature")
plt.title("Weather")

plt.plot(days, max_t, label="Maximum")
plt.plot(days, min_t, label="Minimum")
plt.plot(days, avg_t, label="Average")

plt.legend(loc="upper right",
           shadow=True,
           fontsize="large")

plt.grid(True)
plt.show()
```

# BAR CHARTS USING MATPLOTLIB
### Introduction to Bar Charts
A bar chart is a type of data visualization used to represent categorical data using rectangular bars.

The height or length of each bar represents the corresponding numerical value.

Bar charts are commonly used to compare values between different categories.

Matplotlib provides two main functions for creating bar charts:

+ plt.bar() – Used to create vertical bar charts.
+ plt.barh() – Used to create horizontal bar charts.
#### Simple Bar Chart
A simple bar chart is used to display the revenue of major technology companies.

#### Example:

```python
import matplotlib.pyplot as plt
import numpy as np

company = ["Google", "Amazon", "Flipkart", "Facebook"]

revenue = [90, 136, 89, 27]

a = np.arange(len(company))

plt.bar(a, revenue, label="Revenue")

plt.xticks(a, company)
plt.ylabel("Revenue")
plt.title("Tech Company Stocks")

plt.legend()
plt.show()
```

### Explanation:
+ company – Stores the names of companies.
+ revenue – Stores the revenue values.
+ np.arange() – Generates positions for the bars.
+ plt.bar() – Creates a vertical bar chart.
+ plt.xticks() – Displays company names on the X-axis.
+ plt.ylabel() – Assigns a label to the Y-axis.
+ plt.legend() – Displays the legend.
+ plt.show() – Displays the chart.
### Multiple Bar Charts
Multiple bar charts are used to compare two or more sets of values for different categories.

For example, we can compare the revenue and profit of major technology companies.

### Example:

```python
import matplotlib.pyplot as plt
import numpy as np

company = ["Google", "Amazon", "Flipkart", "Facebook"]

revenue = [90, 136, 89, 27]
profit = [40, 2, 34, 12]

a = np.arange(len(company))

plt.bar(a, revenue, label="Revenue")
plt.bar(a, profit, label="Profit")

plt.xticks(a, company)
plt.ylabel("Amount")
plt.title("Tech Company Stocks")

plt.legend()
plt.show()
```

### Explanation:
+ revenue – Contains the revenue values.
+ profit – Contains the profit values.
+ plt.bar() – Creates bars for revenue and profit.
+ label – Assigns names to the bar groups.
+ plt.legend() – Identifies the different data series.
Note: In this example, the bars overlap because both sets use the same positions.

## Separate Bar Charts (Grouped Bar Charts)
Grouped bar charts display multiple bars side by side for each category.

The width property controls the width of each bar.

### Example:

```python
import matplotlib.pyplot as plt
import numpy as np

company = ["Google", "Amazon", "Flipkart", "Facebook"]

revenue = [90, 136, 89, 27]
profit = [40, 2, 34, 12]

a = np.arange(len(company))

plt.bar(a - 0.2, revenue, width=0.4,
        label="Revenue", color="green")

plt.bar(a + 0.2, profit, width=0.4,
        label="Profit", color="red")

plt.xticks(a, company)
plt.ylabel("Amount")
plt.title("Tech Company Stocks")

plt.legend()
plt.show()
```

### Explanation:
+ a - 0.2 – Positions revenue bars slightly to the left.
+ a + 0.2 – Positions profit bars slightly to the right.
+ width=0.4 – Sets the width of each bar.
+ color="green" – Displays revenue bars in green.
+ color="red" – Displays profit bars in red.
## Horizontal Bar Chart
A horizontal bar chart displays data using horizontal rectangular bars.

In Matplotlib, we use plt.barh() to create a horizontal bar chart.

### Example:

```python
import matplotlib.pyplot as plt
import numpy as np

company = ["Google", "Amazon", "Flipkart", "Facebook"]

revenue = [90, 136, 89, 27]
profit = [40, 2, 34, 12]

a = np.arange(len(company))

plt.barh(a, revenue, label="Revenue")
plt.barh(a, profit, label="Profit")

plt.yticks(a, company)
plt.xlabel("Revenue")
plt.title("Tech Company Stocks")

plt.legend()
plt.show()
```

### Explanation:
+ plt.barh() – Creates horizontal bars.
+ plt.yticks() – Displays company names on the Y-axis.
+ plt.xlabel() – Assigns a label to the X-axis.
+ plt.legend() – Displays the labels of different data series.
Note: The two sets of bars overlap in this example. They can be separated by adjusting their vertical positions.

##  Multiple Bar Charts Using Branch Data
We can also use grouped bar charts to compare the number of students who passed in different branches.

### Example:

```python
import matplotlib.pyplot as plt
import numpy as np

Branch = ["IT", "ECE", "CSE"]

IT = [12, 30, 18, 22, 15]
ECE = [28, 6, 16, 5, 10]
CSE = [29, 3, 24, 25, 17]

a = np.arange(len(IT))

barwidth = 0.25

plt.bar(a - barwidth, IT, color='r',
        width=barwidth, edgecolor='grey',
        label="IT")

plt.bar(a, ECE, color='g',
        width=barwidth, edgecolor='grey',
        label="ECE")

plt.bar(a + barwidth, CSE, color='b',
        width=barwidth, edgecolor='grey',
        label="CSE")

plt.xlabel("Branch")
plt.ylabel("Students Passed")
plt.title("Students Passed in Different Branches")

plt.legend()
plt.show()
```

### Explanation:

+ np.arange() – Generates positions for the bars.
+ barwidth = 0.25 – Sets the width of each bar.
+ a - barwidth – Positions IT bars.
+ a – Positions ECE bars.
+ a + barwidth – Positions CSE bars.
+ color='r' – Red colour for IT.
+ color='g' – Green colour for ECE.
+ color='b' – Blue colour for CSE.
+ edgecolor='grey' – Adds a grey border around the bars.
+ plt.legend() – Displays the branch names.
