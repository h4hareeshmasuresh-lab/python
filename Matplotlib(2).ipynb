# HISTOGRAM
### Aim
To create different types of histograms using Python Matplotlib.

### Introduction
A histogram is one of the most common ways to visualize the distribution of numerical data.

+ X-axis represents the variable or data intervals (bins).
+ Y-axis represents the frequency (number of observations).
### 1. Basic Histogram
Program:

```python
import matplotlib.pyplot as plt

blood_sugar = [
    113, 85, 90, 150, 149, 88, 82,
    129, 93, 115, 135, 80, 77
]

plt.hist(blood_sugar)

plt.title("Blood Sugar Histogram")
plt.xlabel("Blood Sugar Level")
plt.ylabel("Frequency")

plt.show()
```

### 2. Histogram Using Bins and Width
Program:

```python
import matplotlib.pyplot as plt

blood_sugar = [
    113, 85, 90, 150, 149, 88, 82,
    129, 93, 115, 135, 80, 77
]

plt.hist(
    blood_sugar,
    rwidth=0.5,
    bins=5,
    color='yellow'
)

plt.xlabel("Blood Sugar Level")
plt.ylabel("Number of Patients")
plt.title("Blood Sugar Chart")

plt.show()
```

### Explanation:
+ rwidth=0.5 controls the relative width of the bars.
+ bins=5 divides the data into five intervals.
+ color='yellow' sets the bar colour.
### 3. Histogram with Normal, Pre-diabetic and Diabetic Categories
Program:

```python
import matplotlib.pyplot as plt

blood_sugar = [
    113, 85, 90, 150, 149, 88, 82,
    129, 93, 115, 135, 80, 77
]

plt.xlabel("Blood Sugar")
plt.ylabel("Number of Patients")

plt.title("Blood Sugar Chart")

plt.hist(
    blood_sugar,
    rwidth=0.3,
    bins=[80, 100, 125, 150],
    color='b'
)

plt.xticks(
    [90, 112, 137],
    [
        "Normal\n(80-100)",
        "Pre-diabetic\n(100-125)",
        "Diabetic\n(125-150)"
    ]
)

plt.show()
```

### Explanation:
+ bins specifies the interval boundaries.
+ xticks() changes the labels displayed on the X-axis.
+ \n moves text to the next line.
###### Note: These categories are illustrative labels from the practical example, not clinical diagnostic thresholds.

### 4. Multiple Data Samples Histogram
Program:

```python
import matplotlib.pyplot as plt

blood_sugar_men = [
    113, 85, 90, 150, 149, 88, 82,
    129, 93, 115, 135, 80, 77
]

blood_sugar_women = [
    67, 98, 89, 120, 133, 150, 84,
    69, 89, 79, 120, 112, 100
]

plt.xlabel("Blood Sugar")
plt.ylabel("Number of Patients")

plt.hist(
    [blood_sugar_men, blood_sugar_women],
    rwidth=0.3,
    bins=[60, 80, 100, 125, 150],
    color=['yellow', 'red'],
    label=['Men', 'Women']
)

plt.legend()
plt.title("Blood Sugar Comparison")

plt.show()
```

### Explanation:
+ Two datasets are displayed in one histogram.
+ color specifies different colours.
+ label gives names to each dataset.
+ legend() displays the labels.
### 5. Step-Type Histogram
Program:

```python
import matplotlib.pyplot as plt

blood_sugar = [
    113, 85, 90, 150, 149, 88, 82,
    129, 93, 115, 135, 80, 77
]

plt.hist(
    blood_sugar,
    bins=[80, 100, 125, 150],
    histtype='step',
    color='red'
)

plt.xlabel("Sugar Level")
plt.ylabel("Number of Patients")

plt.title("Blood Sugar Chart")

plt.show()
```

### Explanation:
histtype='step' displays the histogram as an outline without filled bars.
### 6. Horizontal Histogram
Program:

```python
import matplotlib.pyplot as plt

blood_sugar = [
    113, 85, 90, 150, 149, 88, 82,
    129, 93, 115, 135, 80, 77
]

plt.hist(
    blood_sugar,
    rwidth=0.4,
    bins=[80, 100, 125, 150],
    color='blue',
    orientation='horizontal'
)

plt.xlabel("Number of Patients")
plt.ylabel("Sugar Level")

plt.title("Blood Sugar Chart")

plt.show()
```

### Explanation:
+ orientation='horizontal' displays the histogram horizontally.
+ X-axis represents frequency.
+ Y-axis represents the sugar level intervals.

# SCATTER PLOT
### Aim
To create different types of scatter plots using Python Matplotlib.

### Introduction
A scatter plot is one of the most common ways to visualize the relationship between two numerical variables.

+ X-axis represents the X values.
+ Y-axis represents the Y values.
+ Each dot represents one observation.
### 1. Basic Scatter Plot
Program:

```python
import matplotlib.pyplot as plt

x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11]

y = [99, 86, 87, 88, 100, 86, 103, 87, 94, 78]

plt.scatter(
    x, y,
    color='red',
    marker='x',
    s=50,
    alpha=0.7
)

plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.title("Basic Scatter Plot")

plt.show()
```

### Explanation:
+ scatter() creates a scatter plot.
+ color='red' sets the dot colour.
+ marker='x' changes the marker shape.
+ s=50 specifies marker size.
+ alpha=0.7 controls transparency.
### 2. Coloured Scatter Plot Based on Values
Program:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(50)
y = np.random.rand(50)

colors = np.random.rand(50)

plt.scatter(
    x, y,
    c=colors,
    cmap='viridis',
    s=100
)

plt.colorbar()

plt.title("Color Mapped Scatter Plot")

plt.show()
```

### Explanation:
+ np.random.rand(50) generates 50 random values.
+ c=colors assigns numerical values to the dots.
+ cmap='viridis' selects a colour map.
+ colorbar() displays the colour scale.
### Other colour maps:
+ plasma
+ inferno
+ magma
+ cividis
+ coolwarm
### 3. Bubble Chart Style Scatter Plot
Program:

```python
import matplotlib.pyplot as plt
import numpy as np

x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11]

y = [99, 86, 87, 88, 100, 86, 103, 87, 94, 78]

sizes = 1000 * np.random.rand(10)

colors = np.random.rand(10)

plt.scatter(
    x, y,
    s=sizes,
    c=colors,
    cmap='plasma',
    alpha=0.6,
    edgecolors='white'
)

plt.colorbar()

plt.title("Bubble Chart Style Scatter Plot")

plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()

```

### Explanation:
+ s=sizes controls the size of each dot.
+ c=colors assigns different colours.
+ cmap='plasma' selects the colour map.
+ alpha=0.6 controls transparency.
+ Larger bubbles represent larger size values.
### 4. Scatter Plot Using Pandas
Program:

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "height": [160, 170, 180, 175, 165],
    "weight": [55, 65, 75, 70, 60]
})

plt.scatter(
    df["height"],
    df["weight"]
)

plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")

plt.title("Height vs Weight")

plt.show()
```

### Explanation:
+ DataFrame() creates a table using Pandas.
+ df["height"] accesses the height column.
+ df["weight"] accesses the weight column.
+ scatter() displays the relationship between height and weight.

# Pie Chart 

### 1. Simple pie chart

```python
import matplotlib.pyplot as plt

exp_vals = [1400, 600, 300, 410, 250]

exp_labels = ["Home rent", "Food", "Bill", "Car", "Other utilities"]

plt.pie(exp_vals, labels=exp_labels)

plt.axis("equal")

plt.show()
```

### 2. Pie chart with percentage, shadow and exploded slices

```python
import matplotlib.pyplot as plt

exp_vals = [1400, 600, 300, 410, 250]

exp_labels = ["Home rent", "Food", "Bill", "Car", "Other utilities"]

plt.pie(exp_vals,
        labels=exp_labels,
        shadow=True,
        autopct='%0.1f%%',
        radius=2,
        explode=[0.5, 0.3, 0.4, 0.2, 0.2])

plt.axis("equal")

plt.show()
```

### 3. Pie chart in clockwise direction

```python
import matplotlib.pyplot as plt

exp_vals = [1400, 600, 300, 410, 250]

exp_labels = ["Home rent", "Food", "Bill", "Car", "Other utilities"]

plt.pie(exp_vals,
        labels=exp_labels,
        shadow=True,
        autopct='%0.1f%%',
        radius=2,
        counterclock=False,
        explode=[0, 0, 0, 0.2, 0.2],
        startangle=90)

plt.axis("equal")

plt.show()
```

### 4. Save pie chart in PDF format

```python
import matplotlib.pyplot as plt

exp_vals = [1400, 600, 300, 410, 250]

exp_labels = ["Home rent", "Food", "Bill", "Car", "Other utilities"]

plt.pie(exp_vals,
        labels=exp_labels,
        shadow=True,
        autopct='%0.1f%%',
        radius=2,
        explode=[0, 0, 0, 0.1, 0.2])

plt.axis("equal")

plt.savefig("expenses-pie.pdf",
            bbox_inches="tight",
            pad_inches=3)

plt.show()
```
