# Type Casting in Python
Type casting is the process of converting one datatype into another datatype to perform a required operation.

There are two types of type conversion:
+ Implicit type conversion
+ Explicit type conversion
### 1. Implicit Type Conversion
In implicit type conversion, Python automatically converts one datatype into another datatype.

For example, when an integer and a float are added, Python converts the integer into a float.

```python
a = 5
b = 3.0

print(type(a))
print(type(b))

c = a + b

print(c)
print(type(c))

d = a * b

print(d)
print(type(d))
```

### 2. Explicit Type Conversion
In explicit type conversion, the programmer manually converts one datatype into another using functions such as:

+ int()
+ float()
+ str()
  
### int()
  The int() function converts a suitable value into an integer.

```python
a = "5"

n = int(a)

print(n)
print(type(n))
```

Note: int("5.9") is not valid because "5.9" is not an integer-formatted string.



### float()
The float() function converts a suitable value into a floating-point number.

Example:

```python
a = "5.9"

n = float(a)

print(n)
print(type(n))
```

### str()
The str() function converts a value into a string.

```python
a = 5

n = str(a)

print(n)
print(type(n))
```

```python
b = "Hello, world!"

print(b[2:])
```

## Set in Python
A set is a collection of unique elements.

Characteristics of Set
+ A set is unordered.
+ A set does not allow duplicate elements.
+ Sets are mutable, meaning elements can be added or removed.
+ Set elements do not support normal indexing.
+ Set elements do not support slicing.
+ Sets are generally represented using curly brackets {}.
+ Sets are useful for mathematical set operations such as union, intersection, and difference.

A set is a collection of elements that is:

Unordered – The elements do not have a fixed position or index.

Unindexed – We cannot access elements using an index.

Mutable – We can add or remove elements.

Unique – Duplicate elements are automatically removed.



```python
s = {'a', 'b', 'c', 'a'}
print(s)
```

The duplicate 'a' is removed automatically.

### Difference Between List and Set
List	           Set

List is ordered.	Set is unordered.

Insertion order is preserved.	No fixed order is guaranteed.

Duplicate elements are allowed.	Duplicate elements are not allowed.

Elements are enclosed in [].	Elements are enclosed in {}.

Indexing is allowed.	Indexing is not allowed.

Slicing is allowed.	Slicing is not allowed.

Example: [1, 2, 3]	Example: {1, 2, 3}

### Creating a Set
Method 1: Using Curly Braces {}

```python
s = {'a', 'b', 'c', 'a'}
print(s)
```

Method 2: Using the set() Function

```python
s = set(('a', 'b', 'c', 'a'))
print(s)
```

Important Note

An empty set cannot be created using {} because {} creates an empty dictionary.

s = set()

This creates an empty set.
### Methods of Sets
#### 1.add()
The add() method is used to add one element to a set.

Syntax

set_name.add(element)

```python
s = {'a', 'b', 'c'}
s.add('d')
print(s)
```

#### 2. update()
The update() method adds multiple elements from another set or other iterable to the existing set.

If an item is present in both sets, only one occurrence will be present because sets do not allow duplicates.

```python
s = {'a', 'b', 'c'}
y = {'a', 'd'}

s.update(y)
print(s)
```

#### 3 pop()
The pop() method removes and returns an arbitrary element from the set.

```python
s = {'a', 'b', 'c'}
s.pop()
print(s)
```

The element removed can vary because a set is unordered.

Important-
Do not assume that pop() always removes the first or last element.

#### 4 remove()
The remove() method removes a specified element from the set.

```python
s = {'a', 'b', 'c'}
s.remove('b')
print(s)
```

Important-
If the specified element does not exist, remove() raises a KeyError.

#### 5 discard()
The discard() method also removes a specified element from a set.

The difference is that discard() does not raise an error if the element does not exist.

```python
s = {'a', 'b', 'c'}
s.discard('a')
print(s)
```

### Set Operations
Set operations are used to compare and combine two or more sets.

Let:

x = {'a', 'b', 'c'}

y = {'b', 'c', 'd'}

#### 1 Union
The union of two sets contains all the elements from both sets.

Syntax

x.union(y)

```python
x = {'a', 'b', 'c'}
y = {'b', 'c', 'd'}

z = x.union(y)
print(z)
```

The duplicate elements 'b' and 'c' appear only once.

Operator

Union can also be performed using |.

z = x | y

#### 2 Intersection
The intersection of two sets contains only the elements that are common to both sets.

```python
x = {'a', 'b', 'c'}
y = {'b', 'c', 'd'}

z = x.intersection(y)
print(z)
```

Operator

Intersection can also be performed using &.

z = x & y

#### 3 Subset – issubset()
The issubset() method checks whether all elements of one set are present in another set.

It returns either True or False.

```python
x = {'a', 'b'}
y = {'a', 'b', 'c'}

z = x.issubset(y)
print(z)
```

This is because every element of x is present in y.

#### 4 Disjoint Set – isdisjoint()
The isdisjoint() method checks whether two sets have no common elements.

It returns True if there are no common elements.

```python
x = {'a', 'b', 'c'}
y = {'m', 'n', 'o'}

z = x.isdisjoint(y)
print(z)
```

There are no common elements between x and y.

#### Example with common elements

```python
x = {'a', 'b', 'c'}
y = {'c', 'd', 'e'}

print(x.isdisjoint(y))
```

#### 5 Difference
The difference() method returns the elements that are present in the first set but not in the second set.

```python
x = {'a', 'b', 'c'}
y = {'a'}

z = x.difference(y)
print(z)
```

Here, 'a' is removed because it is present in both sets.

Operator

Difference can also be performed using -.

z = x - y

## Python Dictionary

A dictionary is a mutable data type in Python that stores elements in the form of key-value pairs.

Each item has a key and a corresponding value.

The key and value are separated by a colon (:).

Multiple key-value pairs are separated by commas (,).

Dictionary elements are enclosed within curly braces { }.

Dictionary keys must be unique.

### Example

```python
d = {
    "name": "Vivek",
    "roll_no": 75,
    "std": 10
}
```

Here:

name, roll_no, and std are keys.

"Vivek", 75, and 10 are their corresponding values.

The dictionary contains 3 key-value pairs.
#### Operations on a Dictionary
Consider the following dictionary:

```python
d = {
    "name": "Vivek",
    "age": 20,
    "height": 160
}
```

Modify an Existing Value

We can modify the value of an existing key by assigning a new value.

```python
d["age"] = 18
print(d)
```

Add a New Key-Value Pair

We can add a new key-value pair to an existing dictionary.

```python
d["dept"] = "Computer"
print(d)
```

 Delete a Key-Value Pair

The del statement can be used to delete a key-value pair.

```python
del d["height"]
print(d)
```

The copy() method returns a copy of the dictionary.

```python
d = {'name': 'Vivek', 'age': 20}

d2 = d.copy()
print(d2)
```

The values() method returns a view containing all the values available in the dictionary.

The keys() method returns a view containing all the keys in the dictionary.

```python
d = {'name': 'Vivek', 'age': 20}

print(d.keys())
```

The items() method returns a view containing the dictionary's key-value pairs as tuples.

Example:

```python
d = {'name': 'Vivek', 'age': 20}

print(d.items())
```

The update() method updates the dictionary with the key-value pairs from another dictionary.

```python
d1 = {'name': 'Vivek', 'age': 20}
d2 = {'course': 'BE'}

d1.update(d2)
print(d1)
```

```python
d = {'name': 'abcd', 'age': 20}

print(d.get('age'))
print(d.get('phone'))
print(d.get('phone', 12345))
```

```python
d = {'brand': 'Ford', 'model': 'Mustang', 'year': 1964}

d.popitem()
print(d)
```

The del keyword is used to delete a dictionary completely or to delete a specific item from a dictionary.

```python
d = {'brand': 'Ford', 'model': 'Mustang'}

del d
print(d)
```

The fromkeys() method returns a new dictionary with the specified keys and a specified value.

Example:

```python
x = ('key1', 'key2', 'key3')
y = 0

d = dict.fromkeys(x, y)
print(d)
```
