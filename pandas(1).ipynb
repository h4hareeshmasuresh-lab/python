## Introduction to Pandas
Pandas is a Python library used for data manipulation and analysis. It provides powerful and flexible tools to clean, explore, organize, and prepare data.

To use Pandas, import the library:

### Creating a Series
A Series is a one-dimensional labeled array capable of holding different data types, such as integers, strings, floats, and Python objects.

#### Creating a Series from a List

```python
import pandas as pd

S1 = pd.Series([10, 20, 30, 40])
print(S1)
```

The index is automatically assigned as [0, 1, 2, 3].

#### Creating a Series with a Custom Index

```python
import pandas as pd

S1 = pd.Series(
    [10, 20, 30, 40],
    index=['a', 'b', 'c', 'd']
)

print(S1)
```

Here, custom labels are assigned to the values instead of the default integer index.

#### Creating a Series from a Dictionary

```python
import pandas as pd

S1 = pd.Series({
    'a': 10,
    'b': 20,
    'c': 30,
    'd': 40
})

print(S1)
```

The dictionary keys become the index labels, and the dictionary values become the Series data.

#### Creating a Series from a Scalar Value
A scalar value is a single value. Pandas repeats it for every index label provided.

```python
import pandas as pd

S1 = pd.Series(5, index=[0, 1, 2, 3])

print(S1)
```

# Replacing Values Using replace()
The replace() method in Pandas is used to substitute specific values with new values in a Series or DataFrame.

## Syntax
#### DataFrame.replace(to_replace, value)
to_replace: The value or values that you want to replace.

value: The new value that replaces the original value.

#### Creating a DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Dave'],
    'Gender': ['F', 'M', 'M', 'unknown'],
    'Score': [85, 90, -1, 95]
})

print(df)
```

###  Replacing a Single Value
We can replace a single value in a particular column using replace().

```python
df['Score'] = df['Score'].replace(-1, 0)
print(df)
```

Explanation: Replaces -1 with 0 in the Score column.

### Replacing Multiple Values with One Value
We can replace multiple values with the same new value.

```python
df['Gender'] = df['Gender'].replace(
    ['F', 'M'], 'known'
)
print(df)
```

Explanation: Replaces both 'F' and 'M' with 'known' in the Gender column.

### Replacing Multiple Values with Different Values
We can use a dictionary to replace different values with different new values.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Dave'],
    'Gender': ['F', 'M', 'M', 'unknown'],
    'Score': [85, 90, -1, 95]
})
df['Gender'] = df['Gender'].replace({
    'F': 'Female',
    'M': 'Male'
})
print(df)
```

### Explanation:
'F' is replaced with 'Female'.

'M' is replaced with 'Male'.

Other values, such as 'unknown', remain unchanged.
### Using Regular Expressions with replace()
Regular expressions (regex) allow us to replace values that match a specified text pattern.

```python
df['Name'] = df['Name'].replace(
    r'^A.*', 'Name starts with A', regex=True
)
print(df)
```

### Explanation:
^A matches text beginning with the letter A.

.* matches any characters after A.

regex=True enables regular-expression matching.

In this example, Alice is replaced with 'Name starts with A'.
### Replacing Values Across the Entire DataFrame
The replace() method can be applied to the entire DataFrame.

```python
df = df.replace('unknown', 'Missing')
print(df)
```

Explanation: Replaces every occurrence of 'unknown' with 'Missing' throughout the DataFrame.

## Renaming Columns Using rename()
The rename() method is used to change column names or index labels in a DataFrame.

### Syntax
df.rename(columns={'old_name': 'new_name'})

#### Important parameters:

+ columns: A dictionary mapping old column names to new column names.
+ inplace=False: Returns a modified DataFrame without changing the original DataFrame directly.
+ inplace=True: Applies the changes directly to the existing DataFrame.

##### Renaming Specific Columns
We can rename selected columns by passing a dictionary to the columns parameter.

```python
import pandas as pd

df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'gender': ['F', 'M', 'F'],
    'score': [85, 90, 95]
})

renamed_df = df.rename(columns={
    'name': 'Name',
    'gender': 'Gender',
    'score': 'Marks'
})

print(renamed_df)
```

### Explanation:

+ 'name' is renamed to 'Name'.
+ 'gender' is renamed to 'Gender'.
+ 'score' is renamed to 'Marks'.
The original DataFrame df remains unchanged because inplace=False is the default.
### Renaming Columns Using inplace=True

```python
df.rename(columns={
    'name': 'Name',
    'gender': 'Gender'
}, inplace=True)

print(df)
```

Explanation: The column names are changed directly in the original DataFrame.

###  Renaming a Single Column

```python
df.rename(columns={'score': 'Marks'}, inplace=True)
```

Explanation: Changes the column name score to Marks.
