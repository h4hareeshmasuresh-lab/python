# Python Functions
####  Introduction to Functions
A function in Python is a block of statements that performs a specific task.

The main idea of using a function is to group commonly used statements together so that the same code can be reused multiple times.

Instead of writing the same code again and again for different inputs, we can define a function once and call it whenever required.

#### Example

```python
def fun():
    print("Hello")

fun()
```

#### Benefits of Using Functions
The main benefits of functions are:

+ Code Reuse – The same code can be used multiple times.
+ Reduced Code Length – We do not need to write the same statements repeatedly.
+ Improved Readability – Functions make the program easier to understand.
+ Easy Maintenance – Changes can be made in one place.
+ Modularity – A large program can be divided into smaller parts.
#### Python Function Definition
A function is defined using the def keyword.

Syntax

def function_name(parameters):

    # statements
    
    return expression
    
Explanation
+ def → keyword used to define a function.
+ function_name → name of the function.
+ parameters → values received by the function.
+ : → indicates the beginning of the function body.
+ return → used to return a value from the function.

# Function with Arguments and No Return Value
A function can receive values through arguments.

#### Example

```python
def add(num1, num2):
    total = num1 + num2
    print(total)

add(15, 30)
```

Here:

+ num1 and num2 are parameters.
+ 15 and 30 are arguments.
+ The function prints the result.
+ The function does not return a value.
## Calling a Function Multiple Times
Once a function is defined, it can be called multiple times.

```python
def fun():
    print("Hello")

fun()
fun()
fun()
```

This demonstrates code reusability.

### Function with an Argument

```python
def world(name):
    print("Hello " + name)

world("Anu")
world("Akhil")
```

The value of name changes according to the argument passed during the function call.

## Function with Multiple Arguments
A function can accept more than one parameter.

```python
def world(name, age):
    print("Hello", name)
    print("Age:", age)

world("Anu", 45)
world("Akhil", 24)
```

### Function with Conditional Statements
A function can contain if, elif, and else statements.

```python
def check_number(num):
    if num > 0:
        print("Number is positive")
    elif num < 0:
        print("Number is negative")
    else:
        print("Number is zero")

check_number(45)
check_number(-3)
check_number(0)
```

####  Default Arguments
A default argument is a parameter that has a default value.

If a value is not provided during the function call, the default value is used.

#### Example

```python
def myfun(x, y=50):
    print("x:", x)
    print("y:", y)

myfun(10)
```

Here, y automatically gets the value 50.

We can also provide another value:

```python
def myfun(x, y):
    print("x:", x)
    print("y:", y)
myfun(10, 20)
```

## Keyword Arguments
In keyword arguments, we specify the parameter name when calling the function.

#### Example


```python
def student(fname, lname):
    print(fname, lname)

student(fname="Geeks", lname="Practice")
student(lname="Practice", fname="Geeks")
```

The order does not matter when keyword arguments are used.

For example:

```python
def student(fname, lname):
    print(fname, lname)
student(lname="Geeks", fname="Practice")
```

### Positional Arguments
In positional arguments, values are assigned to parameters according to their position.

#### Example

```python
def name_age(name, age):
    print("Hi, I am", name)
    print("My age is", age)

name_age("Suraj", 27)
```

The first argument is assigned to name, and the second argument is assigned to age.

If we change the order:

```python
name_age(27, "Suraj")
```

Important Point

In positional arguments, the order of arguments is important.

### Arbitrary Arguments — *args
Sometimes we do not know how many arguments will be passed to a function.

Python provides *args to accept a variable number of positional arguments.

#### Example

```python
def myfun(*args):
    for arg in args:
        print(arg)

myfun("Hello", "Welcome", "to", "Python")
```

args contains all the positional arguments.

args is just a conventional name. The important part is *.
### Arbitrary Keyword Arguments — **kwargs
**kwargs is used to accept a variable number of keyword arguments.

#### Example

```python
def student(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

student(name="Anu", age=20, course="Python")
```

kwargs stores the arguments in dictionary form.

kwargs is a conventional name. The important part is **.
####  Docstring
A docstring is a string written immediately after the function definition.

It is used to describe the purpose or functionality of a function.

The use of a docstring is optional, but it is considered good programming practice.

#### Example

```python
def even_odd(x):
    """Function to check if the number is even or odd."""

    if x % 2 == 0:
        print("Even")
    else:
        print("Odd")

print(even_odd.__doc__)
```

The docstring can be accessed using:

function_name.__doc__

### Nested Functions
A function defined inside another function is called a nested function or inner function.

A nested function can access variables from the enclosing function.

#### Example

```python
def outer():
    s = "I love Python"

    def inner():
        print(s)

    inner()

outer()
```

Here, inner() is defined inside outer().

### Return Statement
The return statement is used to:

Exit from a function.

Send a value back to the calling code.

#### Example

```python
def square_value(num):
    return num ** 2

print(square_value(2))
print(square_value(-4))
```

The function calculates the square and returns the result.



 ## print() vs return

```python
def add(a, b):
    print(a + b)

add(10, 20)
```

Here, the result is only displayed.

### Function Returning Multiple Values
Python allows a function to return multiple values.

#### Example

```python
def mul_div(a, b):
    c = a * b
    d = a / b
    return c, d

x, y = mul_div(20, 10)

print(x)
print(y)
```

Multiple returned values are packed into a tuple.

## Function to Reverse a Number
The following function reverses a given number.

```python
def rev_num(n):
    rev = 0

    while n != 0:
        digit = n % 10
        rev = rev * 10 + digit
        n = n // 10

    return rev

n = int(input("Enter a number: "))

r = rev_num(n)

print("Reverse:", r)
```

#### Explanation
For 12345:

+ Last digit = 5
+ Last digit = 4
+ Last digit = 3
+ Last digit = 2
+ Last digit = 1
Therefore, the reverse is:

54321

## Function to Check Whether a Number is Prime
A prime number is a number greater than 1 that has exactly two factors: 1 and itself.

#### Program

```python
def prime(n):
    if n <= 1:
        print("Not prime")
        return

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            print("Not prime")
            return

    print("Prime")

prime(7)
prime(8)
```

### Function to Find Factorial
The factorial of a number n is represented by n!.

For example:

5! = 5 × 4 × 3 × 2 × 1
   = 120
#### Program

```python
def fact(n):
    factorial = 1

    for i in range(1, n + 1):
        factorial = factorial * i

    print("Factorial ({0}) is {1}".format(n, factorial))

fact(5)
```
