# What is NumPy?
NumPy stands for Numerical Python. It is a Python library used to store and process numerical data efficiently. Its main structure is the ndarray (N-dimensional array). NumPy is commonly used for mathematical calculations, statistics, scientific computing, and as a foundation for libraries such as Pandas and scikit-learn.

### Importing NumPy

import numpy as np

The alias np is a widely used convention. After importing, NumPy functions can be called using np, for example np.array() and np.zeros().


###  Creating NumPy Arrays
Use np.array() to convert a list or nested list into a NumPy array.


```python
import numpy as np
 
list1 = [1, 2, 3, 4]
arr1 = np.array(list1)
print(arr1)

```

A one-dimensional array contains a single sequence of elements. A two-dimensional array contains rows and columns. A three-dimensional array can be thought of as a collection of 2D arrays.

```python
one_d = np.array([1, 2, 3, 4])
two_d = np.array([[1, 2], [3, 4]])
three_d = np.array([[[1, 2], [3, 4]]])
 
print(one_d.ndim)   
print(two_d.ndim)    
print(three_d.ndim)  
```

## Important Array Attributes
Attributes describe an array. They do not normally need parentheses.
+ 	ndim: number of dimensions (axes).
+	size: total number of elements.
+	shape: length along each dimension, returned as a tuple.
+	dtype: data type of the elements


```python
a = np.array([[1, 4], [2, 9], [3, 6], [3, 5]])
print(a.ndim)  
print(a.size)   
print(a.shape)  
print(a.dtype)  # commonly int64 or int32, depending on platform
```

The exact integer dtype may vary by platform and NumPy build.
### Accessing Array Elements (Indexing)
Python uses zero-based indexing: the first element has index 0. Negative indexes count from the end.


```python
arr = np.array([1, 2, 3, 4, 5])
print(arr[0])   
print(arr[-1])  
```

For a 2D array, use row and column indexes. The first index selects the row; the second selects the column.

```python
a = np.array([[1, 4], [2, 9], [3, 6], [3, 5]])
print(a[0])       #  (first row)
print(a[2, 0])    # (third row, first column)
print(a[2, 1])    # (third row, second column)
```

### Array Slicing
Slicing selects a range of elements. The stop index is excluded. The general form is array[start:stop:step].


```python
arr = np.array([10, 20, 30, 40, 50, 60])
print(arr[1:4])  
print(arr[:3])   
print(arr[3:])  
print(arr[::2])  
```

For a 2D array, slicing can be applied to rows and columns, such as arr[1:, :2].

## NumPy Arrays vs Python Lists
Multiplying a Python list by 2 repeats the list. Multiplying a NumPy numeric array by 2 multiplies every element by 2.


```python
items = [1, 2, 3, 4]
arr = np.array(items)
print(items * 2)  
print(arr * 2)    
```

NumPy supports vectorized operations, so calculations can be performed on entire arrays without writing an explicit loop.
# Reshaping and Flattening
reshape() changes the dimensions of an array without changing its data order. The requested shape must contain the same total number of elements.


```python
a = np.array([[1, 4], [2, 9], [3, 6], [3, 5]])  # 8 elements
b = a.reshape(2, 4)
print(b)
```

```python
flat = a.ravel()
print(flat)  
```

ravel() returns a flattened one-dimensional view where possible. flatten() also returns a 1D array but always creates a copy.

# Data Types and dtype Conversion
A NumPy array normally stores elements of one common data type. Examples include integers, floating-point values, and strings.


```python
a = np.array([1, 2, 3])
b = np.array([1.1, 2.1, 3.1])
c = np.array(["a", "b", "c"])
print(a.dtype)
print(b.dtype)
print(c.dtype)
```

Specify a type with dtype= when creating an array, or convert an existing array with astype().

```python
arr = np.array([1, 2, 3], dtype=np.int64)
float_arr = arr.astype(np.float16)
unsigned_arr = arr.astype(np.uint16)
```

Use integer types for whole numbers and floating-point types for decimal values. Unsigned integers cannot represent negative values. Narrow data types have limited ranges, so choose them carefully to avoid overflow or unexpected conversions.

# Mathematical and Statistical Operations
NumPy includes functions for common calculations. These can operate on the whole array or along a selected axis.


```python
a = np.array([[1, 4], [2, 9], [3, 6], [3, 5]])
print(a.sum())   # total of all elements
print(a.min())   # smallest element
print(a.max())   # largest element
print(a.mean())  # arithmetic mean
print(np.sqrt(a)) # square root of each element
```

Understanding axis
For a 2D array, axis=0 operates down the rows and returns one result per column. axis=1 operates across each row and returns one result per row.


```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(np.sum(arr))          
print(np.sum(arr, axis=0))  
print(np.sum(arr, axis=1)) 
print(np.mean(arr, axis=0)) 
print(np.max(arr, axis=1)) 
```

```python
import numpy as np
arr=np.array([[1,2,3],[4,5,6]])
print(np.mean(arr))
print(np.max(arr))
print(np.max(arr,axis=0))
print(np.max(arr,axis=1))
print(np.min(arr))
print(np.min(arr,axis=0))
print(np.min(arr,axis=1))
```

# Filtering with Conditions (Boolean Indexing)
A comparison such as arr > 10 produces a Boolean array containing True and False. Using that Boolean array as an index selects only elements where the condition is True.


```python
arr = np.array([1, 5, 8, 12, 20, 3])
mask = arr > 10
print(mask)       
print(arr[mask]) 
```

Use & for element-wise AND and | for element-wise OR. Put each comparison in parentheses.

```python
arr = np.array([1, 5, 8, 12, 20, 3])
selected = arr[(arr > 5) & (arr < 15)]
print(selected)  
```

Do not use Python keywords and/or for element-wise array conditions; use & and | with parentheses.
#  Element-wise Arithmetic, Dot Product, and Matrix Multiplication
Arithmetic operators such as +, -, *, and / work element by element when array shapes are compatible. The dot product and matrix multiplication are different from element-wise multiplication.


```python
x = np.array([[1, 2], [3, 4]])
y = np.array([[2, 4], [5, 6]])
print(x + y)  # element-wise addition
print(x * y)  # element-wise multiplication
print(np.dot(x, y)) # matrix product for 2D arrays
```

For one-dimensional arrays, the dot product multiplies corresponding values and adds the products.

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.dot(a, b))  # 1*4 + 2*5 + 3*6 = 32
```

For 2D arrays, matrix multiplication can also be written as x @ y. The number of columns in the first matrix must equal the number of rows in the second matrix.

# Inverse of a Matrix
The function np.linalg.inv() calculates the inverse of a square matrix when an inverse exists. A matrix is invertible only when it is non-singular.


```python
a = np.array([[4, 7], [2, 6]], dtype=float)
a_inv = np.linalg.inv(a)
print(a_inv)
print(a @ a_inv)  # approximately the identity matrix
```

If a matrix is singular, NumPy raises a LinAlgError. In practical numerical work, solving a linear system with np.linalg.solve(A, b) is often preferable to explicitly calculating an inverse.

# Matrix Multiplication — np.matmul()
The np.matmul() function is used to multiply two matrices.

#### Example

```python
import numpy as np

M = np.array([[1, 2],
              [3, 4]])

result = np.matmul(M, M)

print(result)
```

#### Explanation:

+ First row, first column: (1 × 1) + (2 × 3) = 7
+ First row, second column: (1 × 2) + (2 × 4) = 10
+ Second row, first column: (3 × 1) + (4 × 3) = 15
+ Second row, second column: (3 × 2) + (4 × 4) = 22
# Cumulative Operations in NumPy
Cumulative operations refer to operations on an array that produce results across elements in a step-by-step fashion.

1. np.cumsum() — Cumulative Sum
Computes the cumulative sum of array elements.

#### Example:

```python
import numpy as np

a = np.array([1, 2, 3, 4])

print(np.cumsum(a))
```

### Explanation:

First element: 1

1 + 2 = 3

3 + 3 = 6

6 + 4 = 10

2. np.cumprod() — Cumulative Product
Computes the cumulative product of array elements.

#### Example:

```python
import numpy as np

a = np.array([1, 2, 3, 4])

print(np.cumprod(a))
```

### Explanation:

First element: 1

1 × 2 = 2

2 × 3 = 6

6 × 4 = 24

3. Cumulative Maximum — np.maximum.accumulate()
Computes the cumulative maximum of array elements.

#### Example:

```python
import numpy as np

a = np.array([2, 1, 5, 3])

print(np.maximum.accumulate(a))
```

### Explanation:

First element: 2

Maximum of (2, 1) = 2

Maximum of (2, 1, 5) = 5

Maximum of (2, 1, 5, 3) = 5

Note: In NumPy versions that support it, np.cummax() can also be used directly.

4. Cumulative Minimum — np.minimum.accumulate()
Computes the cumulative minimum of array elements.

#### Example:

```python
import numpy as np

a = np.array([4, 2, 7, 1])

print(np.minimum.accumulate(a))
```

### Explanation:

First element: 4

Minimum of (4, 2) = 2

Minimum of (4, 2, 7) = 2

Minimum of (4, 2, 7, 1) = 1

Note: In NumPy versions that support it, np.cummin() can also be used directly.

3. Axis Support in Cumulative Operations
All cumulative operations support the axis parameter for multidimensional arrays.

#### Example 1: axis=1
Calculates the cumulative sum across each row.

```python
import numpy as np

b = np.array([[1, 2, 3],
              [4, 5, 6]])

print(np.cumsum(b, axis=1))
```

#### Example 2: axis=0
Calculates the cumulative sum down each column.


```python
import numpy as np

b = np.array([[1, 2, 3],
              [4, 5, 6]])

print(np.cumsum(b, axis=0))
```
