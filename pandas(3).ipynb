# COMBINING DATAFRAMES IN PANDAS
1. CONCAT()
The concat() function is used to combine two or more DataFrames.

It can combine DataFrames in two ways:
+ Vertically (axis=0): Stacks rows one below another.
+ Horizontally (axis=1): Stacks columns side by side.
#### Syntax
pd.concat([df1, df2])

##### Vertical concatenation (default):

pd.concat([df1, df2], axis=0)

##### Horizontal concatenation:

pd.concat([df1, df2], axis=1)
#### Example

```python
import pandas as pd
df1 = pd.DataFrame({
    "a": [1, 2],
    "b": [3, 4]
})
df2 = pd.DataFrame({
    "a": [5, 6],
    "b": [7, 8]
})
result = pd.concat([df1, df2])
print(result)
```

```python
pd.concat([df1, df2], axis=1)
```

```python
pd.concat([df1, df2], axis=0)
```

2. MERGE()
   
The merge() function combines two DataFrames based on common columns (keys).

It is similar to SQL joins.

#### Syntax
pd.merge(df1, df2, on="key")

By default, it performs an INNER JOIN.

#### Types of Merge
1. Inner Join

Returns only matching records from both DataFrames.

pd.merge(df1, df2, on="key", how="inner")

2. Left Join

Returns all records from the left DataFrame and matching records from the right DataFrame.

pd.merge(df1, df2, on="key", how="left")

3. Right Join

Returns all records from the right DataFrame and matching records from the left DataFrame.

pd.merge(df1, df2, on="key", how="right")

4. Outer Join

Returns all records from both DataFrames. Non-matching values are filled with NaN.

pd.merge(df1, df2, on="key", how="outer")

5. Merge Using Different Column Names

When the key column names are different:

pd.merge(
    df1,
    df2,
    left_on="id1",
    right_on="id2"
)
#### Example 1: Inner Join


```python
import pandas as pd

df1 = pd.DataFrame({
    "key": ["a", "b", "c"],
    "value": [1, 2, 3]
})

df2 = pd.DataFrame({
    "key": ["a", "b", "d"],
    "value": [100, 200, 300]
})

result = pd.merge(df1, df2, on="key")

print(result)
```

Note: Only matching keys (a and b) are returned.

#### Example 2: Left Join

```python
import pandas as pd

df1 = pd.DataFrame({
    "key": ["a", "b", "c"],
    "value": [1, 2, 3]
})

df2 = pd.DataFrame({
    "key": ["a", "b", "d"],
    "value": [100, 200, 300]
})
result = pd.merge(
    df1,
    df2,
    on="key",
    how="left"
)

print(result)
```

#### Example 3: Right Join

```python
import pandas as pd

df1 = pd.DataFrame({
    "key": ["a", "b", "c"],
    "value": [1, 2, 3]
})

df2 = pd.DataFrame({
    "key": ["a", "b", "d"],
    "value": [100, 200, 300]
})
result = pd.merge(
    df1,
    df2,
    on="key",
    how="right"
)

print(result)
```

#### Example 4: Merge Using a Common ID

```python
import pandas as pd

df1 = pd.DataFrame({
    "id": [1, 2],
    "name": ["Alice", "Bob"]
})

df2 = pd.DataFrame({
    "id": [1, 2],
    "Score": [90, 85]
})

result = pd.merge(df1, df2, on="id")

print(result)
```

3. JOIN()

The join() function is used to combine columns of another DataFrame using the index or a key column.

By default, join() combines DataFrames using their indexes.

#### Syntax
Join using index:

df1.join(df2)

Join using a key column:

df1.join(
    df2.set_index("key"),
    on="key"
)
#### Example 1: Join Using Index

```python
import pandas as pd

df1 = pd.DataFrame(
    {"name": ["Alice", "Bob"]},
    index=[0, 1]
)

df2 = pd.DataFrame(
    {"Score": [90, 85]},
    index=[0, 1]
)

result = df1.join(df2)

print(result)
```

#### Example 2: Join Using a Key Column

```python
import pandas as pd

df1 = pd.DataFrame({
    "key": [1, 2],
    "name": ["Alice", "Bob"]
})

df2 = pd.DataFrame({
    "key": [3, 4],
    "Score": [90, 85]
})

result = df1.join(
    df2.set_index("key"),
    on="key"
)

print(result)
```

```python
import pandas as pd

df1 = pd.DataFrame({
    "key": [1, 2],
    "name": ["Alice", "Bob"]
})

df2 = pd.DataFrame({
    "key": [3, 4],
    "Score": [90, 85]
})

result = df2.join(
    df1.set_index("key"),
    on="key"
)

print(result)
```

4. APPEND()

The append() method was previously used to add rows from one DataFrame to another.

However, DataFrame.append() was deprecated in pandas 1.4 and removed in pandas 2.0.

The recommended alternative is pd.concat().

Old Method (Not Recommended)

df1.append(df2)
Recommended Method

pd.concat([df1, df2])
#### Example

```python
import pandas as pd

df1 = pd.DataFrame({
    "a": [1, 2],
    "b": [3, 4]
})

df2 = pd.DataFrame({
    "a": [5, 6],
    "b": [7, 8]
})

result = pd.concat(
    [df1, df2],
    ignore_index=True
)

print(result)
```

#### What is ignore_index=True?
ignore_index=True resets the row index after combining DataFrames.

It prevents duplicate index values and creates a new continuous index starting from 0.

#### Important Points to Remember
1. concat() is used for stacking rows or columns.
2. merge() is used to combine DataFrames based on common keys.
3. join() is mainly used for combining DataFrames based on indexes.
4. append() is removed from modern pandas versions.
5. pd.concat() is the recommended alternative to append().
6. NaN represents missing values in a DataFrame.

# DATA TRANSFORMATION IN PANDAS
Applying Functions: apply(), map() and applymap()

In Pandas, functions are used to perform operations on Series and DataFrames.

The three commonly used methods are:
+ apply()
+ map()
+ applymap() (replaced by map() for DataFrames in modern pandas)
1. apply()
The apply() function is used for both Series and DataFrames.

It applies a function to the elements of a Series or along rows or columns of a DataFrame.

A. apply() on Series
Applies a function to each element in the Series.
#### EXAMPLES

```python
import pandas as pd
s = pd.Series([1, 2, 3, 4])
result = s.apply(lambda x: x**2)
print(result)
```

#### Explanation:
+ lambda x: x**2 squares each element.
+ apply() applies this function to every element in the Series.

B. apply() on DataFrame

When used with a DataFrame, apply() applies a function along an axis (rows or columns).

#### Example:

```python
import pandas as pd

df = pd.DataFrame({
    "a": [1, 2, 3],
    "b": [10, 20, 30]
})

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "a": [1, 2, 3],
    "b": [10, 20, 30]
})
df.apply(lambda row: row["a"] + row["b"], axis=1)
```

#### Explanation:
+ axis=1 means row-wise operation.
+ The function adds the values of columns a and b in each row.

Column-wise operation (axis=0)

```python
import pandas as pd

df = pd.DataFrame({
    "a": [1, 2, 3],
    "b": [10, 20, 30]
})
df.apply(lambda col: col.sum(), axis=0)
```

#### Explanation:
+ axis=0 means column-wise operation.
+ The function calculates the sum of each column.
  
Applying a function to a single column

```python
import pandas as pd

df = pd.DataFrame({
    "a": [1, 2, 3],
    "b": [10, 20, 30]
})
df["a"].apply(lambda x: x + 100)
```

#### Explanation:

Adds 100 to every element in column a.

#### Important Points
+ axis=0 → Applies the function to each column.
+ axis=1 → Applies the function to each row.
+ apply() works with both Series and DataFrames.

2. map()
   
The map() function is used for Series only.

It applies a function element-wise to every value in a Series.

It also supports dictionaries and other mapping objects.

#### Example

```python
import pandas as pd

s = pd.Series([1, 2, 3, 4])

result = s.map(lambda x: x * 10)

print(result)
```

#### Explanation:
+ lambda x: x * 10 multiplies each value by 10.
+ map() applies the function to every element.
#### Example using a dictionary

```python
import pandas as pd

s = pd.Series(["a", "b", "c"])

result = s.map({
    "a": "Apple",
    "b": "Banana",
    "c": "Cherry"
})

print(result)
```

### Important Point
map() is used only with Series, not directly with DataFrames.

3. map() ON DATAFRAMES (ELEMENT-WISE)

In modern Pandas, DataFrame.map() is used to apply a function element-wise to every cell in a DataFrame.

The older method DataFrame.applymap() was deprecated in pandas 2.1 and removed in pandas 3.0.

#### Example

```python
import pandas as pd

df = pd.DataFrame({
    "a": [1, 2, 3],
    "b": [10, 20, 30]
})

result = df.map(lambda x: x * 2)

print(result)
```

#### Explanation:
+ DataFrame.map() applies the function to every individual cell.
+ lambda x: x * 2 multiplies every value by 2.
+ Older Syntax (For Reference)

df.applymap(lambda x: x * 2)

Note: Use df.map() in modern pandas versions.

## Easy Way to Remember
+ apply() → Series and DataFrame.
+ Series.map() → Series only.
+ DataFrame.map() → Every cell in a DataFrame.

# PRACTICAL PROBLEMS USING PANDAS
### Problem 1: Calculate Total Marks
Read a CSV file and calculate the total marks of each student.

#### Program:

```python
import pandas as pd
df = pd.read_csv("pd4(3).csv")
df["Total"] = df[
    ["english", "malayalam", "hindi"]
].sum(axis=1)
print(df[["name", "Total"]])
```

### Explanation:
+ read_csv() reads the CSV file.
+ df[["english", "Malayalam", "hindi"]] selects the three subject columns.
+ sum(axis=1) calculates the total marks row-wise.
+ A new column named Total is created.
+ print() displays the student names and total marks.

### Problem 2: Group Students and Calculate Statistics
Group the data by student name and calculate total marks, average marks and number of attempts.

### Program:

```python
import pandas as pd

df = pd.read_csv("pd4(3).csv")

df["Total"] = df[
    ["english", "malayalam", "hindi"]
].sum(axis=1)

result = df.groupby("name")["Total"].agg(
    total_mark="sum",
    average_mark="mean",
    num_attempt="count"
)

print(result)
```

### Explanation:
+ groupby("name") groups records based on student names.
+ ["Total"] selects the Total column.
+ sum calculates the total marks.
+ mean calculates the average marks.
+ count counts the number of attempts.

### Problem 3: Calculate Mean Using groupby()
Calculate the average marks of each student.

### Program:

```python
import pandas as pd

df = pd.read_csv("pd4(3).csv")

result = df.groupby("name").mean(numeric_only=True)

print(result)
```

### Explanation:
+ groupby("name") groups records by student name.
+ mean() calculates the average of numeric columns.
+ numeric_only=True selects numeric columns for the calculation.
