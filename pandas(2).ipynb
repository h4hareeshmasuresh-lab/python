# DATA CLEANING AND PREPROCESSING
### Handling Missing Data in Pandas
Handling missing data is a crucial part of data cleaning in data analytics and data science. 

Missing values can affect the accuracy of analysis and machine learning models.

Pandas provides powerful functions to detect, remove, and fill missing values, such as:

+ isnull() – Detect missing values.
+ dropna() – Remove missing values.
+ fillna() – Fill missing values.
#### Creating a DataFrame

```python
import pandas as pd
import numpy as np

data = {
    'name': ['alice', 'bob', 'charlie', 'david'],
    'age': [25, np.nan, 27, 32],
    'city': ['london', 'usa', 'america', np.nan]
}

df = pd.DataFrame(data)

print(df)
```

Here, NaN represents a missing value.

### Detecting Missing Values Using isnull()
The isnull() function returns a DataFrame of True and False values, indicating whether each value is missing.

### Syntax:

```python
print(df.isnull())
```

### Explanation:
+ True indicates a missing value.
+ False indicates a non-missing value.
####  Removing Missing Values Using dropna()
The dropna() function removes rows or columns containing missing values.

#####  Drop rows containing any missing values

```python
print(df.dropna())
```

### Explanation: 
Rows containing at least one missing value are removed by default.

####  Drop columns containing missing values

```python
print(df.dropna(axis=1))
```

### Explanation:
The axis=1 parameter removes columns containing missing values. The axis=0 default removes rows.

####  Filling Missing Values Using fillna()
The fillna() function replaces missing values with specified values.

#####  Fill missing values with zero

```python
print(df.fillna(0))
```

This replaces all missing values with 0.

###  Fill missing values with a string

```python
print(df.fillna("unknown"))
```

This replaces missing values with the string "unknown".

Note: Filling numeric columns with strings may change the data type of those columns.

### Fill missing numerical values with the column mean

The mean is the average of the available numerical values.

```python
df['age'] = df['age'].fillna(df['age'].mean())
print(df)
```

### Explanation:

The mean age is calculated as:

Mean = (25 + 27 + 32) / 3 = 28

The missing age value is replaced with 28.0. The missing city value remains unchanged.

###  Converting Data to Numeric Format
Sometimes numerical data may be stored as text. The pd.to_numeric() function converts values into numeric format.

```python
import pandas as pd
import numpy as np

data = {
    'name': ['alice', 'bob', 'charlie', 'david'],
    'age': [25, np.nan, 27, 32],
    'city': ['london', 'usa', 'america', np.nan]
}

df = pd.DataFrame(data)
df['age'] = pd.to_numeric(
    df['age'],
    errors='coerce'
)
df['age'] = df['age'].fillna(df['age'].mean())
print(df)
```

### Explanation:

errors='coerce' converts invalid numerical values into NaN.

This helps identify invalid or missing numerical data before preprocessing.
### Saving the Cleaned Data to a CSV File
After cleaning the data, it can be saved to a CSV file using to_csv().

```python
import pandas as pd
import numpy as np

data = {
    'name': ['alice', 'bob', 'charlie', 'david'],
    'age': [25, np.nan, 27, 32],
    'city': ['london', 'usa', 'america', np.nan]
}

df = pd.DataFrame(data)
df['age'] = pd.to_numeric(
    df['age'],
    errors='coerce'
)
df['age'] = df['age'].fillna(df['age'].mean())
df.to_csv('output.csv',index=True)
```

### Explanation:

output.csv is the name of the output file.

index=False prevents Pandas from saving the row index as an additional column.

The saved CSV file can be opened in Excel or imported into another Jupyter Notebook.

## Handling Duplicate Data in Pandas
Duplicate data refers to rows that appear more than once in a dataset. Duplicate records can affect the accuracy of data analysis and should be identified and handled during data preprocessing.

Pandas provides two important functions for handling duplicates:

+ duplicated() – Identifies duplicate rows.
+ drop_duplicates() – Removes duplicate rows.
##  duplicated() Function
The duplicated() function identifies duplicate rows and returns a Boolean Series. It returns True for duplicate occurrences and False for the first occurrence by default.

### Syntax:

df.duplicated(subset=None, keep='first')

Parameters:

+ subset: Specifies the column or columns used to identify duplicates.
+ keep='first': Marks later duplicate occurrences as True.
+ keep='last': Marks earlier duplicate occurrences as True.
+ keep=False: Marks all duplicate occurrences as True.

```python
import pandas as pd

df = pd.DataFrame({
    "a": [1, 2, 2, 3, 3, 3],
    "b": ["a", "b", "b", "c", "c", "c"]
})

print(df.duplicated())
```

#### Explanation:
+ Row 0 is not a duplicate.
+ Row 1 is the first occurrence of the values (2, 'b').
+ Row 2 is a duplicate of row 1.
+ Row 3 is the first occurrence of the values (3, 'c').
+ Rows 4 and 5 are duplicates of row 3.
## drop_duplicates() Function
The drop_duplicates() function removes duplicate rows from a DataFrame.

### Syntax:

```python
df.drop_duplicates(
    subset=None,
    keep='first',
    inplace=False
)
```

#### Parameters:
+ subset: Specifies the columns to check for duplicates.
+ keep='first': Keeps the first occurrence.
+ keep='last': Keeps the last occurrence.
+ inplace=False: Returns a new DataFrame without changing the original DataFrame.
### Example:

```python
import pandas as pd

df = pd.DataFrame({
    "a": [1, 2, 2, 3, 3, 3],
    "b": ["a", "b", "b", "c", "c", "c"]
})

print(df.drop_duplicates())
```

### Explanation:

The duplicate rows are removed, and the first occurrence of each unique row is retained. The original row index is preserved.

# Sorting Data in Pandas
Sorting is an important operation in data cleaning and preprocessing. It helps arrange data in a specific order, making datasets easier to read, analyze, and compare.

Pandas provides two important sorting functions:

+ sort_values() – Sorts a DataFrame or Series based on values.
+ sort_index() – Sorts a DataFrame or Series based on index labels.
###  sort_values() Function
The sort_values() function sorts a DataFrame according to the values in one or more columns.

#### Syntax:
df.sort_values(by='column_name', ascending=True)
##### Parameters:

+ by: Specifies the column name used for sorting.
+ ascending=True: Sorts values in ascending order.
+ ascending=False: Sorts values in descending order.
### Example:

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 20]
})

print(df.sort_values(by='Age'))
```

#### Explanation:

The DataFrame is sorted according to the Age column in ascending order, from the smallest age to the largest age.

#### Sorting in descending order:

```python
print(df.sort_values(by='Age', ascending=False))
```

#### Explanation:
The ages are arranged from the largest to the smallest.

## sort_index() Function
The sort_index() function sorts a DataFrame or Series according to its index labels.

#### Syntax:
df.sort_index(axis=0, ascending=True)
#### Parameters:
+ axis=0: Sorts rows according to their index labels. This is the default.
+ axis=1: Sorts columns according to their column labels.
+ ascending=True: Sorts in ascending order.
+ ascending=False: Sorts in descending order.
### Example:

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 20]
}, index=[2, 0, 1])

print(df.sort_index())
```

#### Explanation:
The rows are arranged according to their index labels in ascending order: 0, 1, and 2.

Notice that the row data remains together; only the order of the rows changes.

Sorting index labels in descending order:


```python
print(df.sort_index(ascending=False))
```

#### Explanation:
The rows are arranged from the highest index label to the lowest.

# Aggregation and Grouping Using groupby()
The groupby() function is one of the most powerful features of Pandas.

It allows you to:
+ Split data into groups.
+ Apply a function to each group.
+ Combine the results into a new output.
### Syntax

df.groupby('column_name')

The groupby() function returns a GroupBy object, which can be used for aggregation and transformation.

```python
import pandas as pd
df = pd.read_csv("pd3(2).csv")
df
```

### Group by a Single Column
We can group the data using a single column, such as Department, and calculate the average salary.

### Python Program

```python
import pandas as pd
df = pd.read_csv("pd3(2).csv")
grouped = df.groupby('department')
print(grouped['salary'].mean())
```

### Explanation:
+ groupby('Department') groups employees by department.
+ ['Salary'] selects the Salary column.
+ .mean() calculates the average salary for each department.
## Grouping Using the Sum Function
The sum() function calculates the total salary of employees in each department.

### Python Program

```python
import pandas as pd
df = pd.read_csv('pd3(2).csv')
grouped = df.groupby('department')['salary'].sum()
print(grouped)
```

### Explanation:

The sum() function adds all the salary values belonging to each department.

## Multiple Aggregations Using agg()
The agg() function is used to perform multiple aggregation operations on a column at the same time.

For example, we can calculate:
+ Mean – Average salary.
+ Max – Highest salary.
+ Min – Lowest salary.
### Python Program

```python
import pandas as pd
df = pd.read_csv('pd3(2).csv')
grouped = df.groupby('department')['salary'].agg(
    ['mean', 'max', 'min']
)
print(grouped)
```

### Explanation:
+ mean calculates the average salary.
+ max returns the highest salary.
+ min returns the lowest salary.
All three operations are performed together.

## Group by Multiple Columns
Pandas also allows us to group data using more than one column.

For example, we can group employees by both Department and Employees.

#### Python Program

```python
import pandas as pd
df = pd.read_csv('pd3(2).csv')
result = df.groupby(
    ['department', 'employee']
)['salary'].sum()
print(result)
```

### Explanation:
+ Department groups employees according to their department.
+ Employees creates additional groups based on employee names.
+ sum() calculates the total salary for each group.
## Using agg() with Custom Functions
The agg() function also allows us to provide custom names for aggregation results.

For example, we can calculate total salary, average salary, and number of employees in each department.

#### Python Program

```python
import pandas as pd
df = pd.read_csv('pd3(2).csv')
result = df.groupby('department').agg(
    total_salary=('salary', 'sum'),
    average_salary=('salary', 'mean'),
    num_employees=('employee', 'count')
)
print(result)
```

### Explanation:
+ total_salary calculates the total salary of each department.
+ average_salary calculates the average salary.
+ num_employees counts the number of employees in each department.
